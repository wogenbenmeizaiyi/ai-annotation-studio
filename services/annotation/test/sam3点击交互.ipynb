```python
from PIL import Image
import numpy as np
image = Image.open("/home/weiyn/projects/ai-annotation-studio-service/test/images/ScreenShot_2025-12-26_151030_700.png")
img_np = np.array(image)

from ultralytics.models.sam import SAM3SemanticPredictor
overrides = dict(
    conf=0.25,
    task="segment",
    mode="predict",
    model="/home/weiyn/projects/ai-annotation-studio-service/models/sam_models/sam3.pt",
    half=True
)
predictor = SAM3SemanticPredictor(overrides=overrides)
predictor.set_image(img_np)
results = predictor(text=["car"])

import json

result = results[0]

if result.masks is None:
    print("没有检测到 mask")
else:
    polygons = []

    for i, contour in enumerate(result.masks.xy):
        # contour 是 numpy array [[x1,y1],[x2,y2],...]
        polygons.append({
            "id": i,
            "points": contour.tolist()
        })

    output = {
        "masks": polygons
    }

    with open("masks.json", "w") as f:
        json.dump(output, f, indent=4)

    print("保存完成")


results[0].show()
```
